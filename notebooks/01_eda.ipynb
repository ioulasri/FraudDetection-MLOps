```python
legitimate = df[df['Class'] == 0]
```

```python
Q1 = legitimate['Amount'].quantile(0.25)
Q3 = legitimate['Amount'].quantile(0.75)
```

```python
IQR = Q3 - Q1
```

```python
outlier_mask = (legitimate['Amount'] < Q1 - 1.5*IQR) | (legitimate['Amount'] > Q3 + 1.5*IQR)
```

```python
outlier_mask.sum()
```

```python
df['hour'] = (df['Time'] / 3600) % 24
```

```python
df.columns
```

```python
df['is_night'] = ((df['hour'] > 22) | (df['hour'] < 6))
```

```python
df['day_sin'] = np.sin(2 * np.pi * df['Time'] / 86400)
df['day_cos'] = np.cos(2 * np.pi * df['Time'] / 86400)
```

```python
df['day_sin']
```

```python
df['log_amount'] = np.log1p(df['Amount'])
df['is_small_amount'] = (df['Amount'] < 5).astype(int)
df['is_large_amount'] = (df['Amount'] > 1000).astype(int)
```

```python
df.head()
```

```python
np.random.seed(42)
```

```python
df['user_id'] = np.random.randint(0, 1000, len(df))
```

```python
rolling_features = df.groupby('user_id')['Amount'].agg([
	('user_transaction_count', 'count'),
	('user_avg_amount', 'mean'),
	('usr_std_amount', 'std'),
	('usr_max_amount', 'max'),
]).reset_index()
```

```python
df = df.merge(rolling_features, on='user_id', how='left')
```

```python
df['amount_ratio'] = df['Amount'] / (df['user_avg_amount'] + 1)
```

```python
df['transactions_per_hour'] = df.groupby(['user_id', 'hour']).cumcount() + 1
```
